Script temperatura

```python
def check_temperature_user_input():
    try:
        # Solicita a temperatura ao usuário
        temp_input = input("Por favor, digite a temperatura atual (em °C): ")
        cpu_temp = float(temp_input)

        # Verifica se a temperatura da CPU foi obtida
        if cpu_temp is not None:
            print("Temperatura atual da CPU: " + str(cpu_temp) + "°C")
            if cpu_temp > 30:
                print("ALERTA: Superaquecimento detectado! Temperatura acima de 30°C.")
            else:
                print("Temperatura normal.")
        else:
            # Esta condição teoricamente não deve ser atingida se o input for um número válido
            print("Não foi possível obter a temperatura. Por favor, insira um número válido.")

    except ValueError:
        print("Erro: Entrada inválida. Por favor, digite um número para a temperatura.")
    except Exception as e:
        # Captura e exibe qualquer outro erro que ocorra
        print("Ocorreu um erro ao verificar a temperatura: " + str(e))

# Chama a função para verificar a temperatura
check_temperature_user_input()
```

Automação de Arquivos


```python
import os
import shutil
import random

# 1. Definir o diretório de origem e destino
source_dir = './arquivos_baguncados'
destination_dir = './arquivos_organizados'

# Criar diretórios se não existirem
os.makedirs(source_dir, exist_ok=True)
os.makedirs(destination_dir, exist_ok=True)

# 2. Criar 10 arquivos diferentes para teste
file_extensions = ['.pdf', '.jpg', '.png', '.txt', '.docx', '.xlsx', '.zip', '.mp3', '.py', '.csv']
file_names = ['documento', 'foto', 'imagem', 'nota', 'relatorio', 'planilha', 'backup', 'musica', 'script', 'dados']

print("Criando arquivos de teste...")
for i in range(10):
    ext = random.choice(file_extensions)
    name = random.choice(file_names) + str(i)
    file_path = os.path.join(source_dir, name + ext)
    with open(file_path, 'w') as f:
        f.write(f"Conteúdo do arquivo {name}{ext}")
    print(f"  Criado: {name}{ext}")

print("\nIniciando a organização dos arquivos...")

# 3. Listar todos os arquivos no diretório de origem
for filename in os.listdir(source_dir):
    file_path = os.path.join(source_dir, filename)

    # Ignorar diretórios, apenas processar arquivos
    if os.path.isfile(file_path):
        # 4. Obter a extensão do arquivo
        # os.path.splitext retorna uma tupla (root, ext)
        _, extension = os.path.splitext(filename)

        # Remover o '.' inicial da extensão
        extension = extension[1:].lower() if extension else 'sem_extensao'

        # 5. Criar o diretório de destino para a extensão
        target_folder = os.path.join(destination_dir, extension)
        os.makedirs(target_folder, exist_ok=True)

        # 6. Mover o arquivo para o diretório de destino
        shutil.move(file_path, os.path.join(target_folder, filename))
        print(f"  Movido: {filename} -> {extension}/")

print("\nOrganização concluída!")
print(f"Arquivos foram organizados de '{source_dir}' para '{destination_dir}'.")
```

Consulta Simples a API


```python
import requests

def consultar_cep(cep):
    """
    Consulta a API ViaCEP para obter informações de endereço para um dado CEP.
    Retorna um dicionário com os dados do endereço ou None em caso de erro.
    """
    # Remove quaisquer caracteres não numéricos do CEP
    cep = ''.join(filter(str.isdigit, cep))

    # Verifica se o CEP possui 8 dígitos
    if len(cep) != 8:
        print("Erro: O CEP deve conter 8 dígitos numéricos.")
        return None

    url = f"https://viacep.com.br/ws/{cep}/json/"

    try:
        # Faz a requisição HTTP para a API
        response = requests.get(url)
        response.raise_for_status() # Lança uma exceção para status de erro (4xx ou 5xx)

        # Converte a resposta JSON para um dicionário Python
        dados_cep = response.json()

        # Verifica se a API retornou um erro (CEP não encontrado, por exemplo)
        if 'erro' in dados_cep and dados_cep['erro']:
            print(f"CEP '{cep}' não encontrado.")
            return None
        else:
            return dados_cep

    except requests.exceptions.RequestException as e:
        # Captura erros de requisição (conexão, timeout, etc.)
        print(f"Erro na requisição à API ViaCEP: {e}")
        return None
    except ValueError as e:
        # Captura erros ao tentar decodificar o JSON (resposta inválida)
        print(f"Erro ao processar a resposta da API: {e}")
        return None
    except Exception as e:
        # Captura quaisquer outros erros inesperados
        print(f"Ocorreu um erro inesperado: {e}")
        return None

def formatar_endereco(dados_cep):
    """
    Formata os dados de endereço em uma string legível.
    """
    if not dados_cep:
        return "Informações de endereço não disponíveis."

    endereco_formatado = f"Logradouro: {dados_cep.get('logradouro', 'N/A')}\n"
    endereco_formatado += f"Complemento: {dados_cep.get('complemento', 'N/A')}\n"
    endereco_formatado += f"Bairro: {dados_cep.get('bairro', 'N/A')}\n"
    endereco_formatado += f"Cidade: {dados_cep.get('localidade', 'N/A')}\n"
    endereco_formatado += f"Estado: {dados_cep.get('uf', 'N/A')}\n"
    endereco_formatado += f"CEP: {dados_cep.get('cep', 'N/A')}"

    return endereco_formatado

# --- Exemplo de Uso ---
if __name__ == '__main__':
    print("--- Consulta de Endereço por CEP (ViaCEP) ---")
    while True:
        cep_input = input("Digite um CEP para consultar (apenas números, ou 'sair' para encerrar): ")
        if cep_input.lower() == 'sair':
            break

        if not cep_input:
            print("Por favor, digite um CEP válido.")
            continue

        dados_encontrados = consultar_cep(cep_input)

        if dados_encontrados:
            print("\n--- Endereço Encontrado ---")
            print(formatar_endereco(dados_encontrados))
            print("--------------------------")
        else:
            print("\nNão foi possível obter informações para o CEP fornecido.")
        print("\n")
```
