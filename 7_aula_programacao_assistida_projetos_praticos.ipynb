Calculadora Simples


```python
# Este programa implementa uma calculadora simples que realiza operações básicas entre um número inicial
# e múltiplos números secundários, permitindo vários cálculos até o usuário decidir encerrar.

while True:
    # 1. Solicita o primeiro número (variável 'a'). Não pode ser vazio.
    # O usuário pode digitar 'sair' (minúsculo ou maiúsculo) para encerrar a calculadora.
    num1_input = input('Digite o primeiro número (ou "sair" para encerrar): ').lower()
    if num1_input == 'sair':
        print("Encerrando a calculadora.")
        break
    try:
        num1 = float(num1_input)
    except ValueError:
        print("Entrada inválida para o primeiro número. Por favor, digite um número válido ou 'sair'.")
        print("-" * 30)
        continue # Reinicia o loop principal para pedir o primeiro número novamente

    # 2. Solicita múltiplos números secundários (variável 'b').
    # O usuário pode pressionar Enter sem digitar nada para finalizar a entrada de números secundários.
    num2_list = []
    print("Digite os números secundários (ou pressione Enter sem digitar nada para finalizar a entrada):")
    while True:
        num_b_str = input(f'  Número {len(num2_list) + 1} para "b": ')
        if num_b_str == '':
            break # Finaliza a entrada de números para 'b'
        try:
            num2_list.append(float(num_b_str))
        except ValueError:
            print("Entrada inválida. Por favor, digite um número válido ou pressione Enter para finalizar.")
            # Continua para a próxima iteração do loop interno para pedir outro número para 'b'

    # Verifica se pelo menos um número secundário foi inserido.
    if not num2_list:
        print("Nenhum número secundário foi inserido. Para realizar uma operação, é necessário pelo menos um número 'b'.")
        print("-" * 30)
        continue # Reinicia o loop principal

    # 3. Solicita a operação desejada.
    opcao = input("Escolha uma operação (S para Soma, SU para Subtração, M para Multiplicação, D para Divisão): ").upper()

    resultado = num1 # Inicializa o resultado com o primeiro número

    # 4. Executa a operação com base na escolha do usuário e nos múltiplos números secundários.
    match opcao:
        case "S": # Soma: num1 + num2_1 + num2_2 + ...
            for n in num2_list:
                resultado += n
            print(f'Soma: {resultado}')
        case "SU": # Subtração: num1 - num2_1 - num2_2 - ...
            for n in num2_list:
                resultado -= n
            print(f'Subtração: {resultado}')
        case "M": # Multiplicação: num1 * num2_1 * num2_2 * ...
            for n in num2_list:
                resultado *= n
            print(f'Multiplicação: {resultado}')
        case "D": # Divisão: num1 / num2_1 / num2_2 / ...
            divisao_por_zero = False
            for n in num2_list:
                if n == 0:
                    print("Erro: Não é possível dividir por zero.")
                    divisao_por_zero = True
                    break # Sai do loop de divisão para evitar erros
                resultado /= n
            if not divisao_por_zero:
                print(f'Divisão: {resultado}')
        case _: # Caso 'default' se nenhuma opção corresponder.
            print("Opção inválida. Por favor, escolha entre S, SU, M ou D.")

    print("-" * 30) # Separador para o próximo cálculo
```

Jogo Geométrico

```python
def jogar():
    # Inicializa a pontuação do jogador
    pontos = 0
    # Mensagens de boas-vindas e instruções do jogo
    print("--- BEM-VINDO AO JOGO DAS FORMAS GEOMÉTRICAS ---")
    print("Instruções: Digite o nome da forma ou o número de lados.")
    print("Pressione ENTER sem digitar nada para encerrar o jogo.\n")

    # Loop principal do jogo, continua até o jogador decidir encerrar
    while True:
        # Solicita a entrada do usuário (número de lados da forma)
        entrada = input("Quantos lados tem a forma que você está pensando? ")

        # Verifica se a entrada está vazia para encerrar o jogo
        if entrada == "":
            break

        # Usa a estrutura 'match' para verificar a entrada do usuário
        match entrada:
            case "3":
                print("Acertou! É um TRIÂNGULO. △")
                pontos += 10 # Adiciona pontos por acerto
            case "4":
                print("Boa! Pode ser um QUADRADO ou RETÂNGULO. □")
                pontos += 10
            case "5":
                print("Isso! É um PENTÁGONO. ⬠")
                pontos += 10
            case "6":
                print("Correto! É um HEXÁGONO. ⬡")
                pontos += 10
            case "0":
                print("Exato! O CÍRCULO não possui lados retos. ○")
                pontos += 10
            case _:
                # Caso a entrada não corresponda a nenhuma forma conhecida
                print("Humm, não conheço essa forma ou o número de lados está incorreto.")

        # Exibe a pontuação atual após cada rodada
        print(f"Pontuação atual: {pontos}\n")

    # Mensagem final e pontuação total quando o jogo é encerrado
    print(f"\nJogo encerrado! Sua pontuação final foi: {pontos}")

# Inicia a execução do jogo
jogar()
```

API

```python
import requests # Importa a biblioteca 'requests' para fazer requisições HTTP a APIs externas.

def consultar_cep(cep):
    """
    Consulta a API ViaCEP para obter informações de endereço para um dado CEP.
    Retorna um dicionário com os dados do endereço ou None em caso de erro.
    """
    # Remove quaisquer caracteres não numéricos do CEP (garante que só números sejam processados).
    cep = ''.join(filter(str.isdigit, cep))

    # Verifica se o CEP, após a limpeza, possui exatamente 8 dígitos.
    if len(cep) != 8:
        print("Erro: O CEP deve conter 8 dígitos numéricos.") # Mensagem de erro para CEPs inválidos.
        return None # Retorna None indicando falha na validação.

    # Constrói a URL da API ViaCEP com o CEP fornecido.
    url = f"https://viacep.com.br/ws/{cep}/json/"

    try:
        # Faz a requisição HTTP GET para a URL da API.
        response = requests.get(url)
        # Verifica se a requisição foi bem-sucedida (código de status 200). Se não, lança um erro HTTP.
        response.raise_for_status() # Lança uma exceção para status de erro (4xx ou 5xx).

        # Converte a resposta JSON da API para um dicionário Python.
        dados_cep = response.json()

        # Verifica se a API retornou um campo 'erro', indicando que o CEP não foi encontrado.
        if 'erro' in dados_cep and dados_cep['erro']:
            print(f"CEP '{cep}' não encontrado.") # Mensagem para CEP não encontrado.
            return None # Retorna None para CEPs não encontrados.
        else:
            return dados_cep # Retorna o dicionário com os dados do endereço.

    except requests.exceptions.RequestException as e:
        # Captura e trata erros relacionados à requisição HTTP (ex: problemas de conexão, timeout).
        print(f"Erro na requisição à API ViaCEP: {e}")
        return None
    except ValueError as e:
        # Captura e trata erros ao tentar decodificar o JSON (ex: resposta inválida da API).
        print(f"Erro ao processar a resposta da API: {e}")
        return None
    except Exception as e:
        # Captura quaisquer outros erros inesperados que possam ocorrer durante o processo.
        print(f"Ocorreu um erro inesperado: {e}")
        return None

def formatar_endereco(dados_cep):
    """
    Formata os dados de endereço em uma string legível.
    """
    # Verifica se há dados de CEP para formatar.
    if not dados_cep:
        return "Informações de endereço não disponíveis." # Retorna uma mensagem se não houver dados.

    # Constrói a string do endereço formatado, usando .get() para evitar erros se uma chave não existir.
    endereco_formatado = f"Logradouro: {dados_cep.get('logradouro', 'N/A')}\n"
    endereco_formatado += f"Complemento: {dados_cep.get('complemento', 'N/A')}\n"
    endereco_formatado += f"Bairro: {dados_cep.get('bairro', 'N/A')}\n"
    endereco_formatado += f"Cidade: {dados_cep.get('localidade', 'N/A')}\n"
    endereco_formatado += f"Estado: {dados_cep.get('uf', 'N/A')}\n"
    endereco_formatado += f"CEP: {dados_cep.get('cep', 'N/A')}"

    return endereco_formatado # Retorna a string formatada do endereço.

# --- Exemplo de Uso ---
# Bloco de código que será executado apenas quando o script for rodado diretamente.
if __name__ == '__main__':
    print("--- Consulta de Endereço por CEP (ViaCEP) ---")
    # Loop infinito para permitir múltiplas consultas até que o usuário decida sair.
    while True:
        # Solicita ao usuário que digite um CEP ou 'sair' para encerrar.
        cep_input = input("Digite um CEP para consultar (apenas números, ou 'sair' para encerrar): ")
        # Converte a entrada para minúsculas e verifica se o usuário quer sair.
        if cep_input.lower() == 'sair':
            break # Sai do loop se o usuário digitar 'sair'.

        # Verifica se a entrada do usuário não está vazia.
        if not cep_input:
            print("Por favor, digite um CEP válido.") # Mensagem para entrada vazia.
            continue # Pula para a próxima iteração do loop.

        # Chama a função para consultar o CEP fornecido.
        dados_encontrados = consultar_cep(cep_input)

        # Verifica se a consulta retornou dados (ou seja, o CEP foi encontrado e processado).
        if dados_encontrados:
            print("\n--- Endereço Encontrado ---")
            # Imprime o endereço formatado usando a função formatar_endereco.
            print(formatar_endereco(dados_encontrados))
            print("--------------------------")
        else:
            # Mensagem exibida se não foi possível obter informações para o CEP.
            print("\nNão foi possível obter informações para o CEP fornecido.")
        print("\n") # Imprime uma linha em branco.
```
